```python
from pathlib import Path
import pandas as pd
```

```python
DATA_ROOT = Path(
    "data/idx_financial_statements"
)

print(DATA_ROOT.resolve())
print("Exists:", DATA_ROOT.exists())
```

```python
folders = []

for path in DATA_ROOT.rglob("*"):

    if not path.is_dir():
        continue

    relative_path = path.relative_to(
        DATA_ROOT
    )

    folders.append({
        "folder_name": path.name,
        "relative_path": str(relative_path),
        "depth": len(relative_path.parts),
        "parent_folder": path.parent.name
    })

folder_df = pd.DataFrame(folders)

print(
    "Total folders:",
    len(folder_df)
)

display(
    folder_df.head(100)
)
```

```python
for depth in sorted(
    folder_df["depth"].unique()
):

    print("\n" + "=" * 70)
    print("DEPTH:", depth)
    print("=" * 70)

    names = (
        folder_df[
            folder_df["depth"] == depth
        ]["folder_name"]
        .drop_duplicates()
        .head(100)
    )

    for name in names:
        print(name)
```

```python
COMPANY_DEPTH = 2

company_list = (
    folder_df[
        folder_df["depth"] == COMPANY_DEPTH
    ]["folder_name"]
    .drop_duplicates()
    .sort_values()
    .reset_index(drop=True)
)

company_df = pd.DataFrame({
    "ticker": company_list
})

print(
    "Total unique companies:",
    len(company_df)
)

display(company_df)
```
