```python
from pathlib import Path
import pandas as pd
import numpy as np
import re
```

```python
INVENTORY_FILE = Path(
    "data/idx_dataset_inventory.csv"
)

DATA_ROOT = Path(
    "data/idx_financial_statements"
)

print("Inventory exists:", INVENTORY_FILE.exists())
print("Data root exists:", DATA_ROOT.exists())
```

```python
inventory_df = pd.read_csv(
    INVENTORY_FILE
)

print(
    "Total files:",
    len(inventory_df)
)

display(
    inventory_df.head()
)
```

```python
xlsx_df = inventory_df[
    inventory_df["extension"]
    .str.lower()
    .eq(".xlsx")
].copy()

print(
    "Total XLSX files:",
    len(xlsx_df)
)

display(
    xlsx_df.head(20)
)
```

```python
def extract_ticker(file_name):
    if pd.isna(file_name):
        return None

    file_name = str(file_name)

    match = re.match(
        r"^([A-Z0-9]+)_",
        file_name.upper()
    )

    if match:
        return match.group(1)

    return None


xlsx_df["ticker"] = (
    xlsx_df["file_name"]
    .apply(extract_ticker)
)

display(
    xlsx_df[
        [
            "ticker",
            "year",
            "quarter",
            "file_name",
            "file_path"
        ]
    ].head(30)
)
```

```python
print(
    "Ticker detected:",
    xlsx_df["ticker"]
    .notna()
    .sum()
)

print(
    "Ticker missing:",
    xlsx_df["ticker"]
    .isna()
    .sum()
)

display(
    xlsx_df[
        xlsx_df["ticker"].isna()
    ].head(20)
)
```

```python
TEST_TICKERS = [
    "AADI",
    "BBRI",
    "AMRT",
    "TLKM",
    "KLBF"
]

test_files_df = (
    xlsx_df[
        xlsx_df["ticker"]
        .isin(TEST_TICKERS)
    ]
    .sort_values(
        [
            "ticker",
            "year",
            "quarter"
        ],
        ascending=[
            True,
            False,
            False
        ]
    )
)

display(
    test_files_df[
        [
            "ticker",
            "year",
            "quarter",
            "file_name",
            "file_path"
        ]
    ]
)
```

```python
MIN_VALID_SIZE_MB = 0.01  # sekitar 10 KB

valid_test_files = (
    test_files_df[
        test_files_df["size_mb"] >= MIN_VALID_SIZE_MB
    ]
    .sort_values(
        ["ticker", "year", "quarter"],
        ascending=[True, False, False]
    )
)

latest_test_files = (
    valid_test_files
    .groupby(
        "ticker",
        as_index=False
    )
    .first()
)

display(
    latest_test_files[
        [
            "ticker",
            "year",
            "quarter",
            "file_name",
            "size_mb",
            "file_path"
        ]
    ]
)
```

```python
def resolve_file_path(path_text):
    path = Path(path_text)

    if path.exists():
        return path

    alternative = Path.cwd() / path

    if alternative.exists():
        return alternative

    return None


latest_test_files[
    "resolved_path"
] = (
    latest_test_files[
        "file_path"
    ].apply(resolve_file_path)
)

display(
    latest_test_files[
        [
            "ticker",
            "file_name",
            "resolved_path"
        ]
    ]
)
```

```python
for _, row in latest_test_files.iterrows():

    print(
        row["ticker"],
        "->",
        row["resolved_path"],
        "| exists:",
        row["resolved_path"] is not None
    )
```

```python
failed_excel_files = []

for _, row in latest_test_files.iterrows():

    ticker = row["ticker"]
    file_path = row["resolved_path"]

    print("\n" + "=" * 80)
    print("TICKER:", ticker)
    print("FILE:", file_path)
    print("=" * 80)

    try:
        excel_file = pd.ExcelFile(
            file_path,
            engine="openpyxl"
        )

        for i, sheet in enumerate(
            excel_file.sheet_names
        ):
            print(
                i,
                "-",
                sheet
            )

    except Exception as e:

        print(
            "FAILED TO READ:",
            type(e).__name__,
            "-",
            e
        )

        failed_excel_files.append({
            "ticker": ticker,
            "file_path": str(file_path),
            "error_type": type(e).__name__,
            "error": str(e)
        })
```

```python
failed_excel_df = pd.DataFrame(
    failed_excel_files
)

print(
    "Total failed files:",
    len(failed_excel_df)
)

display(
    failed_excel_df
)
```

```python
for _, row in latest_test_files.iterrows():

    file_path = row["resolved_path"]

    print(
        row["ticker"],
        "| size:",
        file_path.stat().st_size,
        "bytes"
    )
```

```python
amrt_row = latest_test_files[
    latest_test_files["ticker"] == "AMRT"
].iloc[0]

amrt_path = amrt_row["resolved_path"]

with open(amrt_path, "rb") as f:
    raw = f.read()

print("Size:", len(raw), "bytes")
print("Raw content:")
print(raw)
```
